```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("Z:\Internship\Tasks\Edge Detection\image\img3.png")
kernel=np.ones((2,2))
```

```python
erosion=cv2.erode(img,kernel,iterations=1)
dilation=cv2.dilate(img,kernel,iterations=1)
```

```python
edge=dilation-erosion
```

```python
plt.imshow(img,cmap='grey')
plt.show()

plt.imshow(edge,cmap='grey')
plt.show()
```
