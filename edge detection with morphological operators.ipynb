```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("Z:/Internship/Tasks/Edge Detection/image/img2jpg")
img1=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
blurred=cv2.GaussianBlur(img1,(3,3),0)
kernel=np.ones((2,2)) 
```

```python
erosion=cv2.erode(blurred,kernel,iterations=1)
dilation=cv2.dilate(blurred,kernel,iterations=1)
```

```python
edge=dilation-erosion 
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
plt.imshow(edge,cmap='grey')
plt.show()
```

```python
edge1=blurred-erosion
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()

plt.imshow(edge1,cmap='grey')
plt.show()
```

```python
edge2=dilation-blurred
_, binary_edges = cv2.threshold(edge2,10, 100, cv2.THRESH_BINARY)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()

plt.imshow(binary_edges,cmap='grey')
plt.show()
```
