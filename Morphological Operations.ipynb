```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

image = cv2.imread("Z:\Internship\Tasks\Edge Detection\pic6.jpeg")

kernel = np.ones((2,2))

erosion = cv2.erode(image, kernel, iterations=1)
erosion = cv2.erode(erosion, kernel, iterations=1)

dilation = cv2.dilate(erosion, kernel, iterations=1)
# dilation = cv2.dilate(dilation, kernel, iterations=1)
 
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)
 
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)

titles = ['Original Image', 'Erosion', 'Dilation', 'Opening', 'Closing']
images = [image, erosion, dilation, opening, closing]

for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])

plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

image = cv2.imread("Z:\Internship\Tasks\Edge Detection\pic6.jpeg")

blurred=cv2.GaussianBlur(image,(3,3),0)

kernel = np.ones((1,1))

a,binary=cv2.threshold(blurred,128,255,cv2.THRESH_BINARY)

opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
opening = cv2.morphologyEx(opening, cv2.MORPH_OPEN, kernel)


closing = cv2.morphologyEx(opening, cv2.MORPH_CLOSE, kernel)

plt.imshow(closing,cmap='gray')
plt.show()
```
