```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Canny
image1=cv2.imread("Z:\Internship\Tasks\Edge Detection\pic1.jpeg")
image = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
blurred_image=cv2.GaussianBlur(image,(9,9),0)
edges=cv2.Canny(blurred_image, 0,50)

plt.title('image before')
plt.imshow(image1)
plt.show()

plt.title('after image')
plt.imshow(edges, cmap='Blues')
plt.show()   
```

```python
#Sobel
image1=cv2.imread("Z:\Internship\Tasks\Edge Detection\pic1.jpeg")
image = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
blurred_image=cv2.GaussianBlur(image,(9,9),0)
edge2=cv2.Sobel(src=blurred_image,ddepth=cv2.CV_64F,dx=1,dy=1)


plt.title('image before')
plt.imshow(image1)
plt.show()

plt.title('after image')
plt.imshow(edge2, cmap='gray')
plt.show()
```

```python
#Scharr
image1=cv2.imread("Z:\Internship\Tasks\Edge Detection\pic1.jpeg")
image = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
blurred_image=cv2.GaussianBlur(image,(9,9),0)
edge2=cv2.Scharr(src=blurred_image,ddepth=cv2.CV_64F,dx=1,dy=0)

plt.title('image before')
plt.imshow(image1)
plt.show()

plt.title('after image')
plt.imshow(edge2, cmap='gray')
plt.show()
```

```python
x=np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
y=np.array([[-1,-1,-1],[0,0,0],[1,1,1]])
```

```python
#Prewitt
image1=cv2.imread("Z:\Internship\Tasks\Edge Detection\pic1.jpeg")
image = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
blurred_image=cv2.GaussianBlur(image,(9,9),0)
priwittx=cv2.filter2D(blurred_image,cv2.CV_64F,x)
priwitty=cv2.filter2D(blurred_image,cv2.CV_64F,y)

edge3=np.sqrt(priwittx**2,priwitty**2)

plt.title('image before')
plt.imshow(image1) 
plt.show()

plt.title('after image')
plt.imshow(edge3, cmap='gray')
plt.show() 
```

```python
#Laplace
laplace=cv2.Laplacian(blurred_image,cv2.CV_64F)
plt.title('image before')
plt.imshow(image1) 
plt.show()

plt.title('after image')
plt.imshow(laplace, cmap='gray')
plt.show() 
```
